```python
from pyspark import functions as F
```

```python
df = spark.read.csv('../pandas/tips.csv', header=True, inferSchema=True)
```

```python
df = df.withColumn( 'valor_total', F.col('total_bill') + F.col('tip') )
```

```python
resumo = ( df.groupBy('day', 'sex') .agg( F.sum('valor_total').alias('total_vendas'), F.avg('valor_total').alias('media_vendas'), F.count('*').alias('quantidade') ) )
```
