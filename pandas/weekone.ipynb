```python
import pandas as pd
```

# Padronização

```python
df = pd.read_csv('tips.csv')
```

# Nova coluna

```python
df['valor_total'] = df['total_bill'] + df['tip']
```

# Agregações

```python
resumo =(
    df.groupby(['day', 'sex']).agg( total_vendas=('valor_total', 'sum'), 
                                   media_conta=('valor_total', 'mean'), 
                                   quantidade=('valor_total', 'count') ).reset_index() )
```

```python
resumo.to_parquet('resumo.parquet', index=False)
```
